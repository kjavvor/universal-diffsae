```python
import random
import os

def generate_diffusion_prompts(num_prompts, output_filepath):
    """
    Generuje losowe, bardziej różnorodne prompty dla modelu dyfuzyjnego
    i zapisuje je do pliku tekstowego.

    Args:
        num_prompts (int): Liczba promptów do wygenerowania.
        output_filepath (str): Pełna ścieżka do pliku tekstowego,
                                gdzie zostaną zapisane prompty.
    """

    # Znacznie rozszerzone słowniki z kategoriami słów
    subjects = [
        "red car", "blue bicycle", "vintage motorcycle", "futuristic hovercraft",
        "cat", "dog", "owl", "wolf", "fox", "bear", "lion", "tiger", "elephant",
        "dolphin", "whale", "octopus", "squirrel", "rabbit", "deer", "eagle",
        "astronaut", "knight", "wizard", "mermaid", "robot", "cyborg", "samurai",
        "viking", "pirate", "fairy", "elf", "dwarf", "dragon", "unicorn", "griffin",
        "phoenix", "golem", "ghost", "alien", "tree", "flower", "mushroom", "cactus",
        "waterfall", "mountain", "volcano", "river", "lake", "ocean", "desert",
        "forest", "jungle", "cave", "island", "cityscape", "village", "ancient ruin",
        "spaceship", "airship", "submarine", "train", "hot air balloon", "castle",
        "lighthouse", "windmill", "pyramid", "temple", "library", "laboratory",
        "observatory", "art gallery", "bakery", "coffee shop", "bookstore",
        "old man", "young girl", "punk rocker", "business person", "scientist",
        "musician", "dancer", "chef", "gardener", "detective", "superhero",
        "invisible creature", "sentient plant", "time traveler"
    ]
    actions_or_states = [
        "on a sunny day", "sitting on a chair", "riding a horse", "flying high",
        "standing tall", "blooming brightly", "exploring new worlds",
        "sleeping soundly", "running fast", "floating gently", "shining brightly",
        "glowing softly", "singing loudly", "dancing freely", "building a sandcastle",
        "reading a book", "playing a game", "cooking a meal", "painting a picture",
        "writing a story", "meditating peacefully", "jumping with joy",
        "whispering secrets", "observing the stars", "repairing a device",
        "brewing a potion", "crafting an artifact", "chasing butterflies",
        "listening to music", "dreaming vividly", "conducting an experiment",
        "performing a magic trick", "searching for treasure", "escaping from danger",
        "training for a marathon", "contemplating the universe", "weaving a spell",
        "telling a joke", "discovering a relic", "navigating a maze",
        "communicating with animals", "carving a sculpture", "composing a symphony",
        "solving a mystery", "unraveling an ancient scroll", "defying gravity"
    ]
    locations_or_modifiers = [
        "on Mars", "in a deep forest", "by the shimmering sea", "in a bustling cyberpunk city",
        "under the milky way stars", "at a breathtaking sunset", "in a torrential rain",
        "during a fierce thunderstorm", "in a peaceful moonlit meadow",
        "on a desolate alien planet", "inside a chaotic futuristic laboratory",
        "on a crisp autumn morning", "in a hidden crystal cave",
        "next to a thundering waterfall", "on a perpetually busy street",
        "in a dusty, forgotten library", "on top of a towering skyscraper",
        "in a whimsical magical kingdom", "deep underwater in a coral reef",
        "in a vast, arid desert", "among ancient ruins", "on a floating island",
        "inside a colossal spaceship", "beneath a bioluminescent canopy",
        "within a collapsing star", "on the edge of a black hole",
        "in a vibrant neon alley", "on a windswept cliff", "by a frozen lake",
        "through a swirling vortex", "in a parallel dimension",
        "under a rainbow waterfall", "on a gigantic mushroom",
        "in a cloud city", "within a giant treehouse", "at the bottom of the ocean",
        "on a volcanic moon", "in a field of giant sunflowers",
        "during an aurora borealis display", "on a forgotten battlefield",
        "in a bustling marketplace", "at a grand ball", "in a haunted mansion",
        "on a pirate ship", "in a secret garden", "on a cosmic playground",
        "within a dreamscape", "at the dawn of time", "in a post-apocalyptic wasteland"
    ]

    generated_prompts = []
    for _ in range(num_prompts):
        subject = random.choice(subjects)
        action_or_state = random.choice(actions_or_states)
        location_or_modifier = random.choice(locations_or_modifiers)

        # Zwiększona różnorodność w strukturze promptu
        rand_val = random.random()
        if rand_val < 0.6:  # 60% szans na pełny prompt
            prompt = f"A {subject} {action_or_state} {location_or_modifier}"
        elif rand_val < 0.85: # 25% szans na prompt bez lokalizacji/modyfikatora
            prompt = f"A {subject} {action_or_state}"
        else: # 15% szans na prompt "subject with location"
            prompt = f"A {subject} {location_or_modifier}"
        generated_prompts.append(prompt)

    # Zapisywanie promptów do pliku
    try:
        # Upewnienie się, że katalog istnieje
        os.makedirs(os.path.dirname(output_filepath) or '.', exist_ok=True)
        with open(output_filepath, 'w', encoding='utf-8') as f:
            for prompt in generated_prompts:
                f.write(prompt + '\n')
        print(f"Wygenerowano i zapisano {num_prompts} promptów do: {output_filepath}")
    except IOError as e:
        print(f"Błąd podczas zapisu pliku: {e}")

liczba_promptow = 10000 

sciezka_do_pliku = 'big_prompts.txt'

# Wywołanie funkcji generującej prompty
generate_diffusion_prompts(liczba_promptow, sciezka_do_pliku)
```
